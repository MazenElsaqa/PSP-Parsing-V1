# 03 - Table Experiments

نقارن أكثر من إعداد لاستخراج الجداول ونختار الأفضل حسب **دقة البنية** (fill ratio, spans, headers)، مش السرعة.

```python
from pathlib import Path
import pandas as pd
from psp_parsing.pipeline import run_pipeline, compare_table_runs

PDF = (Path('..') / 'data' / 'input' / 'instrumentation-and-control-engineering.pdf').resolve()
root = (Path('..') / 'artifacts' / 'experiments').resolve()
runs = []
for config in ['baseline', 'tables-fast']:
    cfg = (Path('..') / 'configs' / f'{config}.yaml').resolve()
    runs.append(run_pipeline(PDF, cfg, root / config))
rows = compare_table_runs(runs, root / 'table-comparison.json')
pd.DataFrame(rows).sort_values(['table_id', 'run'])
```

قارن `fill_ratio` و`spanning_cells` بين accurate و fast لكل `table_id`. اختر الإعداد اللي يحافظ على الخلايا المدمجة والعناوين، وسجّل السبب في ملاحظاتك.
