نشغّل الـ pipeline الكاملة مرة واحدة وننتج canonical document + tables + images + QA. ده أول تسليم للبيانات العادية اللي نقدر نقارن عليه بعدين.

```python
from pathlib import Path
from psp_parsing.pipeline import run_pipeline

PDF = Path('..') / 'data' / 'input' / 'instrumentation-and-control-engineering.pdf'
CONFIG = Path('..') / 'configs' / 'baseline.yaml'
output_dir = run_pipeline(PDF.resolve(), CONFIG.resolve(), Path('..').resolve() / 'artifacts')
output_dir
```

```python
from psp_parsing.models import CanonicalDocument

document = CanonicalDocument.model_validate_json((output_dir / 'document.json').read_text())
print('elements:', len(document.elements))
headings = [e for e in document.elements if e.heading_level is not None]
for heading in headings[:15]:
    print('  ' * (heading.heading_level - 1), f'H{heading.heading_level}', heading.text[:70])
```
