شغّل من جذر المشروع: `uv run jupyter lab`.

```python
from pathlib import Path
from pypdfium2 import PdfDocument
from psp_parsing.ids import document_id

PDF = Path('..') / 'data' / 'input' / 'instrumentation-and-control-engineering.pdf'
pdf = PdfDocument(PDF)
print('document_id:', document_id(PDF))
print('pages:', len(pdf))
```

```python
# النص الخام لأول صفحتين لفهم بنية العنوان والفهرس
for index in range(min(2, len(pdf))):
    page = pdf[index]
    text = page.get_textpage().get_text_range()
    print(f'--- page {index + 1} ---')
    print(text[:800])
    print()
```
